# Welcome to Data Visualization

```python
import pandas as pd
pd.plotting.register_matplotlib_converters()
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
fifa_data = pd.read_csv(r"C:\Kaggle\Seaborn\fifa.csv", index_col='Date', parse_dates=True)
# Raw String r"" will tell python treat everything as string.
# parse_dates=True tells pandas to convert date/time columns into datetime objects
```

```python
fifa_data.head()
```

```python
plt.figure(figsize=(16,6))
# set the width and height of figure(blank canvas)
sns.lineplot(data=fifa_data)
# creating linechart and it only work with numerical columns in dataframe.
```

# Line Charts

```python
spotify_filepath = "C:\Kaggle\Seaborn\spotify.csv"

spotify_data = pd.read_csv(spotify_filepath, index_col="Date", parse_dates=True)
spotify_data.head()
```

```python
spotify_data.tail()
```

```python
# LineChart
sns.lineplot(data=spotify_data)
```

```python
plt.figure(figsize=(14,6))

# Adding title
plt.title("Daily Global Streams of Popular Songs in 2017-2018")

sns.lineplot(data=spotify_data)
```

```python
# converting dataframe columns into list.
list(spotify_data.columns)
```

```python
plt.figure(figsize=(14,6))
plt.title("Daily Global Streams of Popular Songs in 2017-2018")

# Line Chart only "Shape of You"
sns.lineplot(data=spotify_data['Shape of You'], label="Shape of You")

#Line Chart only of "Despacito"
sns.lineplot(data=spotify_data['Despacito'], label="Despacito")

# label for horizontal axis not mandatory but recommended.
plt.xlabel("Date")
```
