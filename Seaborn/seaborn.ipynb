# Welcome to Data Visualization

```python
import pandas as pd
pd.plotting.register_matplotlib_converters()
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
fifa_data = pd.read_csv(r"C:\Kaggle\Seaborn\fifa.csv", index_col='Date', parse_dates=True)
# Raw String r"" will tell python treat everything as string.
# parse_dates=True tells pandas to convert date/time columns into datetime objects
```

```python
fifa_data.head()
```

```python
plt.figure(figsize=(16,6))
# set the width and height of figure(blank canvas)
sns.lineplot(data=fifa_data)
# creating linechart and it only work with numerical columns in dataframe.
```

# Line Charts

```python
spotify_filepath = "C:\Kaggle\Seaborn\spotify.csv"

spotify_data = pd.read_csv(spotify_filepath, index_col="Date", parse_dates=True)
spotify_data.head()
```

```python
spotify_data.tail()
```

```python
# LineChart
sns.lineplot(data=spotify_data)
```

```python
plt.figure(figsize=(14,6))

# Adding title
plt.title("Daily Global Streams of Popular Songs in 2017-2018")

sns.lineplot(data=spotify_data)
```

```python
# converting dataframe columns into list.
list(spotify_data.columns)
```

```python
plt.figure(figsize=(14,6))
plt.title("Daily Global Streams of Popular Songs in 2017-2018")

# Line Chart only "Shape of You"
sns.lineplot(data=spotify_data['Shape of You'], label="Shape of You")

#Line Chart only of "Despacito"
sns.lineplot(data=spotify_data['Despacito'], label="Despacito")

# label for horizontal axis not mandatory but recommended.
plt.xlabel("Date")
```

# Bar Charts and Heatmaps

```python
flight_data = pd.read_csv(r"C:\Kaggle\Seaborn\flight_delays.csv", index_col="Month")
flight_data
```

```python
plt.figure(figsize=(14,6))
plt.title('Average Arrival Delay for Spirit Airlines')
plt.ylabel('Arrival Delay(in minutes)')
# Bar chart in which first argument is index and second is data.
sns.barplot(x=flight_data.index, y=flight_data['NK'])
```

```python
# Heatmap
plt.figure(figsize=(14,7))

plt.title('Average Arrival Delay for Each Airline, by Month')

sns.heatmap(data=flight_data, annot=True)

plt.xlabel('Airline')
# annot=True will show the values in each cell.
```

# Scatter Plots

```python
insurance_data = pd.read_csv(r"C:\Kaggle\Seaborn\insurance.csv")
insurance_data
```

```python
plt.figure(figsize=(14,8))
sns.scatterplot(x=insurance_data['bmi'], y=insurance_data['charges'])
```

```python
# adding Regression line
plt.figure(figsize=(14,8))
sns.regplot(x=insurance_data['bmi'], y=insurance_data['charges'])
```

## Color-coded Scatter plots

```python
plt.figure(figsize=(14,8))
sns.scatterplot(x=insurance_data['bmi'], y=insurance_data['charges'], hue=insurance_data['smoker'])
# basically we can also able to use three varaibles for scatter plots.
```

```python
# creating two regression line
sns.lmplot(x='bmi', y='charges', hue='smoker', data=insurance_data)
```

## Categorical Scatter Plot

```python
sns.swarmplot(x=insurance_data['smoker'], y=insurance_data['charges'])
```
