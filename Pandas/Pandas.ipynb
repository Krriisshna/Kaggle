## Maps

### map() method

```python
# map() applies a function to each value (like a for loop).
review_points_mean = reviews.points.mean()
reviews.points.map(lambda p: p - review_points_mean)
# lambda is short, anonymous(no name) function.
```

### apply() method

```python
# apply(): applies a function to each row (axis=1) or column (axis=0) as a Series
def remean_points(row):
    row.points = row.points - review_points_mean
    return row

reviews.apply(remean_points, axis='columns')
# map(): value-wise on a single Series | apply(): row-wise or column-wise on Series/DataFrame
# map(): always returns Series
# apply(): returns Series or DataFrame depending on function output
```

## Built-in methods for the same

```python
reviews.points - review_points_mean
# here we have a series in RHS and one value in LHS so python will automatically 
# figures out what to do.
```

```python
reviews.country + '-' + reviews.region_1
# these operations are faster the map() and apply().
# can perform all standard python operations(>, <, ==, ....)
# but map() and apply() can do advanced things.
```

# Grouping and Sorting

## groupby() operation

```python
# groupby() creates groups using the group key (values of the grouped column) and maps them to the corresponding index labels.
# It also sort the group keys.
reviews.groupby('points').points.count()
# it is doing the same work as value_counts() function.
```

```python
# we can use any summary functions with it.
reviews.groupby('points').price.min()
```

```python
# we can think each group as slice of dataframe.
# apply() method can be applied to dataframes.
reviews.groupby('winery').apply(lambda df: df.title.iloc[0])
```

```python
# we can also group more than one column.
reviews.groupby(['country','province']).apply(lambda df: df.loc[df.points.idxmax()])
# idxmax() returns the index label of the first occurrence of the maximum value.
```

```python
# agg() is a GroupBy method used to apply aggregation functions to each group.
reviews.groupby('country').price.agg([len,min,max])
```

## Multi-Index

```python
countries_reviewed = reviews.groupby(['country','province']).description.agg([len])
countries_reviewed
```

```python
mi = countries_reviewed.index
type(mi)
```

```python
# reset_index() method to get regular index
countries_reviewed.reset_index()
```

## Sorting

```python
# Use pandas sort_values() to sort a DataFrame or Series.
countries_reviewed = countries_reviewed.reset_index()
countries_reviewed.sort_values(by='len')
```

```python
# sorting in descending order.
countries_reviewed.sort_values(by='len', ascending=False)
```

```python
# Sorting by index values.
countries_reviewed.sort_index()
```

```python
# sorting via more than one column.
countries_reviewed.sort_values(['country','len'])
```

## Combining

```python
canadian_youtube = pd.read_csv('C:\Kaggle\Pandas\CAvideos.csv')
british_youtube = pd.read_csv('C:\Kaggle\Pandas\GBvideos.csv')

pd.concat([canadian_youtube, british_youtube])
# concat() combines DataFrames along an axis and aligns data by labels.
```

```python
left = canadian_youtube.set_index(['title', 'trending_date'])
right = british_youtube.set_index(['title', 'trending_date'])

left.join(right, lsuffix='_CAN', rsuffix='_UK')
# this will combine data frame based on matching index labels.
```
