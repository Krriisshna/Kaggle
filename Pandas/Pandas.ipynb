# Indexing, Selecting & Assigning

```python
reviews = pd.read_csv('C:\Kaggle\Pandas\winemag-data-130k-v2.csv', index_col=0)
reviews
```

```python
reviews.country # Attribute selection
# accessing country property(column) of reviews.
```

```python
reviews['country']
# indexinng([]) operator works the same but it can handle complex column name.
```

```python
reviews['country'][0]
# accessing a single value of series.
```

## Pandas own accessor operator loc and iloc

### iloc operator

```python
# iloc follow index-based selection(numerical position)
reviews.iloc[0] # select 1st row.
```

```python
# loc & iloc are row-first, column-second opposite of native python.
reviews.iloc[:,0] #accessing column & : means everything
```

```python
reviews.iloc[:3,0] # 1st, 2nd & 3rd row of column.
```

```python
reviews.iloc[1:3,0]
```

```python
reviews.iloc[[1,2,3],0] # list is allowed
```

```python
reviews.iloc[-5:] # 5th row from end to end row.
```

### loc operator

```python
# loc follow label-based selection.
reviews.loc[0,'country']
# but numerical value is working as row index label is numerical.
```
