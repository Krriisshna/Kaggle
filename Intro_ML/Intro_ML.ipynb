# Machine Learning
- Machine learning is the process where computers use mathematical methods to learn patterns from data.
- It uses past data to learn patterns, and then uses those patterns to make predictions on new data.

## Decision Tree
- A Decision Tree is a Machine Learning model that makes decisions by asking a series of questions, like a flowchart.
- It splits data into branches based on conditions to reach a final prediction.
- They are basic building block from some of the best models in datascience.
